```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.dates as mdates

df=pd.read_csv("/content/Temperature Data.csv")
print(df.head())
# we can convert the combined year and month into two different columns as year and month
df['Date'] = pd.to_datetime(df['Date'], format='%Y%m')
print(df.head())
# Make year and month as separate columns
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
print(df.head())
print(df.tail())

# 1. Time Series Analysis Plot of Monthly average temperatures
plt.figure(figsize=(30, 6))
plt.plot(df['Date'],df['Value'],color = 'blue', marker='o', linestyle='-',markersize=3)
plt.title('Monthly Average Temperatures from 1895-2024')
plt.xlabel('Year')
plt.ylabel('Average Temperature (°F)')
plt.grid(True)
locator = mdates.YearLocator(5)  # Set ticks every 5 years
formatter = mdates.DateFormatter('%Y')  # Format ticks as years
ax = plt.gca()  # Get the current axis
ax.xaxis.set_major_locator(locator)
ax.xaxis.set_major_formatter(formatter)
plt.xticks(rotation=90)
plt.show()

#2. Seasonal Pattern Analysis using bar plot
Monthly_avg = df.groupby('Month')['Value'].mean().reset_index()
print(Monthly_avg)
plt.figure(figsize=(20, 6))
plt.bar(Monthly_avg['Month'], Monthly_avg['Value'], color='blue')
plt.title('Seasonal Patterns and Variations')
plt.xlabel('Month')
plt.ylabel('Average Temperature (°F)')
plt.xticks(Monthly_avg['Month'])
plt.show()


#3 Temperature anomalies over time and Histogram Vis

plt.figure(figsize=(10, 6))
plt.plot(df['Date'], df['Anomaly'], color='blue')
plt.title('Temperature Anomalies Over Time (1895-2024)')
plt.xlabel('Year')
plt.ylabel('Anomaly (°F)')
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))
plt.hist(df['Anomaly'], bins=30, color='Red', edgecolor='black')
plt.title('Distribution of Temperature Anomalies (1895-2024)')
plt.xlabel('Anomaly (°F)')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()

# 4. Trend Analysis - Moving average to smooth short-term fluctuations
df['Moving_Avg'] = df['Value'].rolling(window=12).mean()  # 12-month moving average
print(df.head(15))
plt.figure(figsize=(26, 6))
plt.plot(df['Date'], df['Value'], label='Original',color = 'blue')
plt.plot(df['Date'], df['Moving_Avg'], color='orange', label='12-Month Moving Average')
plt.title('Moving Average of Monthly Temperatures')
plt.xlabel('Year')
plt.ylabel('Average Temperature (°F)')
plt.legend()
plt.grid(True)
plt.show()


#5

max_anomaly = df['Anomaly'].max()
min_anomaly = df['Anomaly'].min()
max_anomaly_year = df.loc[df['Anomaly'].idxmax(), 'Date']  # Get the corresponding year for the max anomaly
min_anomaly_year = df.loc[df['Anomaly'].idxmin(), 'Date']
print(max_anomaly_year)
print(min_anomaly_year)

plt.figure(figsize=(26, 10))
plt.plot(df['Date'], df['Anomaly'], color='purple')
plt.title('Extreme Temperature Anomalies')
plt.xlabel('Year')
plt.ylabel('Anomaly (°F)')
plt.grid(True)
# Annotating extreme events of 2012 and 1936.
plt.annotate('Extreme High Temparature Than Normal',xy = (max_anomaly_year,max_anomaly),arrowprops=dict(facecolor='red', shrink=0.05),fontsize=12)
plt.annotate('Extreme low Temparature Than Normal',xy = (min_anomaly_year,min_anomaly),arrowprops=dict(facecolor='blue',shrink = 0.05),fontsize = 12)
plt.show()


#6

decades = {
'1895s': list(range(1895, 1905)),
'1905s': list(range(1905, 1915)),
'1915s': list(range(1915, 1925)),
'1925s': list(range(1925, 1935)),
'1935s': list(range(1935, 1945)),
'1945s': list(range(1945, 1955)),
'1955s': list(range(1955, 1965)),
'1965s': list(range(1965, 1975)),
'1975s': list(range(1975, 1985)),
'1985s': list(range(1985, 1995)),
'1995s': list(range(1995, 2005)),
'2005s': list(range(2005, 2015)),
'2015s': list(range(2015, 2025))
}

decade_wise_temparature={}

for decade, years in decades.items():
    decade_wise_temparature[decade] = df[df['Year'].isin(years)]['Value'].values

print(decade_wise_temparature)

plt.figure(figsize=(12, 8))
plt.boxplot([temps for temps in decade_wise_temparature.values()], labels=decade_wise_temparature.keys())
plt.title('Temperature Distribution Across Decades')
plt.xlabel('Decade')
plt.ylabel('Temperature (°F)')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()


#7

decades = {
    '1895s': list(range(1895, 1905)),
    '1905s': list(range(1905, 1915)),
    '1915s': list(range(1915, 1925)),
    '1925s': list(range(1925, 1935)),
    '1935s': list(range(1935, 1945)),
    '1945s': list(range(1945, 1955)),
    '1955s': list(range(1955, 1965)),
    '1965s': list(range(1965, 1975)),
    '1975s': list(range(1975, 1985)),
    '1985s': list(range(1985, 1995)),
    '1995s': list(range(1995, 2005)),
    '2005s': list(range(2005, 2015)),
    '2015s': list(range(2015, 2025))
}

def assign_decade(year):
    for decade, years in decades.items():
        if year in years:
            return decade
    return None

df['Decade'] = df['Year'].apply(assign_decade)

# Group by 'Decade' and 'Month', and calculate the Avg anomaly for each month
heatmap_data = df.groupby(['Decade', 'Month'])['Anomaly'].mean().unstack()

# Create the heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(heatmap_data, cmap='coolwarm', annot=True, fmt=".2f", cbar_kws={'label': 'Temperature Anomaly (°F)'})
plt.title('Heatmap of Monthly Temperature Anomalies Across Decades')
plt.xlabel('Month')
plt.ylabel('Decade')
plt.show()


#8

plt.figure(figsize=(10, 6))
plt.hist(df['Value'], bins=30, color='blue', edgecolor='black')
plt.title('Frequency Distribution of Monthly Average Temperatures')
plt.xlabel('Average Temperature (°F)')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()


#9

decades = {
    '1895s': list(range(1895, 1905)),
    '1905s': list(range(1905, 1915)),
    '1915s': list(range(1915, 1925)),
    '1925s': list(range(1925, 1935)),
    '1935s': list(range(1935, 1945)),
    '1945s': list(range(1945, 1955)),
    '1955s': list(range(1955, 1965)),
    '1965s': list(range(1965, 1975)),
    '1975s': list(range(1975, 1985)),
    '1985s': list(range(1985, 1995)),
    '1995s': list(range(1995, 2005)),
    '2005s': list(range(2005, 2015)),
    '2015s': list(range(2015, 2025))
}

# Calculate average temperature anomalies for each decade
decade_avg_anomalies = {}

for decade, years in decades.items():
    decade_avg_anomalies[decade] = df[df['Year'].isin(years)]['Anomaly'].mean()

print(decade_avg_anomalies)

# Plot the bar chart of average anomalies across decades
plt.figure(figsize=(12, 8))
plt.bar(decade_avg_anomalies.keys(), decade_avg_anomalies.values(), color='blue')
plt.title('Average Temperature Anomalies Across Decades')
plt.xlabel('Decade')
plt.ylabel('Average Anomaly (°F)')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```
