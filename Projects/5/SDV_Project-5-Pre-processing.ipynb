```python
columns_to_include = [
    'eventid', 'iyear', 'imonth', 'iday', 'country', 'country_txt', 'region_txt', 
    'city', 'success', 'attacktype1_txt', 'nkill', 'propvalue', 'targtype1_txt', 
    'latitude', 'gname', 'longitude', 'targsubtype1_txt', 'target1', 
    'weaptype1_txt', 'weapdetail'
]
data = data[columns_to_include]

column_rename_map = {
    'eventid': 'Event_id',
    'iyear': 'Year',
    'imonth': 'Month',
    'iday': 'Day',
    'country': 'Country_no',
    'country_txt': 'Country',
    'region_txt': 'Region',
    'city': 'City',
    'propvalue': 'property_value',
    'attacktype1_txt': 'Attack_type',
    'targtype1_txt': 'Target_type',
    'gname': 'group_name',
    'targsubtype1_txt': 'Target_sub_type',
    'target1': 'Target',
    'nkill': 'no_of_kills',
    'weaptype1_txt': 'Weapon_type',
    'weapdetail': 'Weapon_detail'
}

data.rename(columns=column_rename_map, inplace=True)

# Save the selected columns to a new CSV file
data.to_csv("F:/UNT_Classes/Semester-4/Scientific Data Visualization/Projects/5/preprocessed_selected_renamed.csv", index=False)
```
