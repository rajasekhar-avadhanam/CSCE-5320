## Part 1
* Use pandas to read the `bestsellers` dataset into a DataFrame
* Once you've done that, use Pandas to figure out how many rows and columns the DF has
* Inspect the first 5 rows
* Inspect the first 19 rows
* Inspect the last 5 rows
* Inspect the last 2 rows
* Which columns (if any) are missing values?
* What datatype did Pandas assign to "User Rating"?
* How many integer columns are in the DataFrame?

```python
import pandas as pd
#Use pandas to read the bestsellers dataset into a DataFrame
df_bs = pd.read_csv('/content/bestsellers.csv')
#Once you've done that, use Pandas to figure out how many rows and columns the DF has
print(df_bs.shape)
#Inspect the first 5 rows
print(df_bs.head(5))
#Inspect the first 19 rows
print(df_bs.head(19))
#Inspect the last 5 rows
print(df_bs.tail(5))
#Inspect the last 2 rows
print(df_bs.tail(2))
#Which columns (if any) are missing values?
msng_val=df_bs.isnull().sum()
print(msng_val)
#What datatype did Pandas assign to "User Rating"?
print(df_bs['User Rating'].dtype)
#How many integer columns are in the DataFrame?
# Count the number of columns with integer data types
int_columns = df_bs.select_dtypes(include='int').columns
print(len(int_columns))
```

## Part 2

* The `mount_everest_deaths` dataset has its own index column provided in the dataset.  When importing it, use the existing index column.
* Which columns have zero null values?
* Which column has the most null values?


```python
import pandas as pd
df_mt = pd.read_csv('/content/mount_everest_deaths.csv',index_col=0)
df_mt.head()
nl_columns = df_mt.isnull().sum()[df_mt.isnull().sum() == 0]
print(nl_columns)
# Find the column with the most null values
most_nl_columns = df_mt.isnull().sum().idxmax()
print(most_nl_columns)
```

* Import the `movie_titles.tsv` dataset
